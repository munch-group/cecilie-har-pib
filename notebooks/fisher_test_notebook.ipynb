## Import packages and load data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import geneinfo.plot as gplt
import geneinfo.information as gi
from geneinfo.genelist import GeneListCollection 
from geneinfo.genelist import GeneList as glist
import geneinfo.ontology as go

```

```python
LinARs_df = pd.read_csv("../data/adc9507_Data_S1.csv", sep=";", skiprows=2)
LinARs_df = LinARs_df.dropna(axis=1, how="all")

# Replace "06.sep" with "SEPT6"
LinARs_df['Gene_name'] = LinARs_df['Gene_name'].replace('06.sep', 'SEPT6')
sep_genes = {
    '05.sep': 'SEPT5',
    '06.sep': 'SEPT6',
    '09.sep': 'SEPT9',
    '14.sep': 'SEPT14',
    '11.sep': 'SEPT11',
    '07.sep': 'SEPT7',
    '03.sep': 'SEPT3',
    '01.sep': 'SEPT1',
    '04.sep': 'SEPT4',
    '08.sep': 'SEPT8',
}
LinARs_df['Gene_name'] = LinARs_df['Gene_name'].replace(sep_genes)

LinARs_df.head()
```

## Fisher test of genes 

### Fisher test (all primates)

```python
# Fisher test (all genes)

# Genes associated with LinARs on chr X
chrX_linars_genes = filtered_LinARs_df[filtered_LinARs_df["Chr"] == "chrX"]["Gene_name"].unique().tolist() 
chrX_linars_genes = glist(chrX_linars_genes) 

# Gene lists 
neuron = glist(neuron_genes_list)
spermatid = glist(human_spermatid_expressed_genes_list)
cDEG = glist(cDEG_human_list)

# Background genes from BioMart (all genes on chrX)
chrx_genes_df = pd.read_csv("../data/chrx_genes.txt", sep="\t")
chrx_genes_list = chrx_genes_df["Gene name"].dropna().unique().tolist()
chrx_genes_list = glist(chrx_genes_list) 



# Run Fisher test for neuron genes
result_neuron = chrX_linars_genes.fisher(neuron, background=chrx_genes_list, return_counts=True) 
print("Neuron result:", result_neuron)
#pvalue: 7.106207256257531e-186 -> huge enrichment
#444 LinAR-associated genes on chrX are neuron genes
#115 LinAR-associated genes on chrX are not neuron genes
#151 non-LinAR chrX genes are neuron genes
#1185 non-LinAR chrX genes are not neuron genes
#Removed 18 genes not in background set


# Run Fisher test for spermatid genes
result_spermatid = chrX_linars_genes.fisher(spermatid, background=chrx_genes_list, return_counts=True) 
print("Spermatid result:", result_spermatid)

# Run Fisher test for cDEG genes
result_cDEG = chrX_linars_genes.fisher(cDEG, background=chrx_genes_list, return_counts=True) 
print("cDEG result:", result_cDEG)

```

```python
# Fisher test (all protein coding genes) 

# Background from BioMart (all protein coding genes on chrX)
chrx_genes_protein_df = pd.read_csv("../data/chrx_protein_coding_genes.txt", sep="\t")
chrx_genes_protein_list = chrx_genes_protein_df["Gene name"].dropna().unique().tolist()
chrx_genes_protein_list = glist(chrx_genes_protein_list)

# Filter chrX_linars_genes to only include genes in chrx_genes_protein_df
chrX_linars_genes_protein = chrX_linars_genes & chrx_genes_protein_list

# Filter neuron, spermatid, and cDEG lists to only include genes in chrx_genes_protein_df
neuron_protein = neuron & chrx_genes_protein_list
spermatid_protein = spermatid & chrx_genes_protein_list
cDEG_protein = cDEG & chrx_genes_protein_list

# Run Fisher test for neuron genes
result_neuron_protein = chrX_linars_genes_protein.fisher(neuron_protein, background=chrx_genes_protein_list, return_counts=True) 
print("Neuron protein coding result:", result_neuron_protein)

# Run Fisher test for spermatid genes
result_spermatid_protein = chrX_linars_genes_protein.fisher(spermatid_protein, background=chrx_genes_protein_list, return_counts=True) 
print("Spermatid protein coding result:", result_spermatid_protein)

# Run Fisher test for cDEG genes
result_cDEG_protein = chrX_linars_genes_protein.fisher(cDEG_protein, background=chrx_genes_protein_list, return_counts=True) 
print("cDEG protein coding result:", result_cDEG_protein)

```
